```python
def asteroidCollision(asteroids: list[int]) -> list[int]:
    a = asteroids
    n = len(a)
    s = []
    c = list(tuple(a))
    
    for i in range(n):
        if c[i] < 0:
            continue
        
        if i + 2 < n:
            if c[i] > -c[i+1]:
                c[i+1] = 0
                if c[i] == -c[i+2]:
                    c[i] = 0
                    c[i+2] = 0
                elif c[i] > -c[i+2]:
                    c[i+2] = 0
                else:
                    c[i] = 0
        if i + 1 < n:
            if c[i] == -c[i+1]:
                c[i] = 0
                c[i+1] = 0
            elif c[i] < -c[i+1]:
                c[i] = 0

    print(c)

asteroidCollision([5,10,-5])

# 1 == -2 = 1, -2 D
# 1 < -2 = 1 D
# 1 > -2, 1 == -3 = 1, -2, -3 D
# 1 > -2, 1 > -3 = -2, -3 D
# 1 > -2, 1 < -3 = -2, 1 D
# 2 == 3 = 2, 3
# 2 > 3 = 3
# 2 < 3 = 2
```
