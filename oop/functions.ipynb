#### global vs nonlocal vs local variables

for `global` variables, this variable is set in the file itself and explicitly mentioning it in the function will ensure that the functions works with this variable

for `nonlocal` variable, this variable is from the nearest enclosing function, the variable cannot be in the file

```python
my_glob = 0

def do_something(value):
    global my_glob
    my_glob += value
    my_nonlocal = 1
    print(f"my_nonlocal before: {my_nonlocal}")
    def inner_func(value):
        nonlocal my_nonlocal
        global my_glob
        my_glob += value
        my_nonlocal += value
    inner_func(88)
    print(f"my_nonlocal after: {my_nonlocal}")

print(f"my_glob before: {my_glob}")
do_something(99)
print(f"my_glob after: {my_glob}")
```
