```python
class MyIter:
    def __init__(self, start, end):
        self.start = start
        self.value = self.start
        self.end = end
    
    # Must have __iter__ if not it will not work
    def __iter__(self):
        return self
    
    def __next__(self):
        if self.value <= self.end:
            v = self.value
            self.value += 2
            return v
        else:
            raise StopIteration

my_iter = MyIter(3,9)
for i in my_iter:
    print(i)   
```
