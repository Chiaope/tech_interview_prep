#### `__new__` vs `__call__`


Using `__new__` will cause `__init__` to always run even though there is already an instance
This will be an issue overriding old data even though there is only 1 instance

Using `__call__` will not cause `__init__` to run

`__call__` will only be run when the class is being called here `abc = MyClass()`, overriding `__call__` can override the default function of creating the class

#### MetaClass
MetaClass is a class that defines how the actual class is being created, not the instance.
example
```python
class MyClass(metaclass=MyMetaClass): # here MyMetaClass will run and do something
```

````python
class MyBasicSingleton:
    _instances = {}

    def __new__(cls, *args, **kwargs):
        if cls not in cls._instances:
            cls._instances[cls] = super().__new__(cls)
        return cls._instances[cls]

    def __init__(self, updated="old"):
        self.updated = updated
        print(f"updated: {self.updated}")
````

````python
aaa = MyBasicSingleton("aaa")
bbb = MyBasicSingleton("bbb")

# old data is overriden by the new 'singleton' even though there can only be 1 instance
aaa.updated
````

````python
# MUST PUT IN `type` EXPLICTLY
class SingletonMetaclass(type):
    _instances = {}
    def __call__(cls, *args, **kwds):
        if cls not in cls._instances:
            cls._instances[cls] = super().__call__(*args, **kwds)
        return cls._instances[cls]

class MySingleton(metaclass=SingletonMetaclass):
    def __init__(self, updated):
        self.updated = updated

class MyAnotherSingleton(metaclass=SingletonMetaclass):
    def __init__(self, updated):
        self.updated = updated
````

````python
class_1 = MySingleton("class_1")
class_2 = MySingleton("class_2")
a_class_1 = MyAnotherSingleton("a_class_1")
a_class_2 = MyAnotherSingleton("a_class_2")
print(f"""
class_1 is class_2: {class_1 is class_2}
a_class_1 is a_class_2: {a_class_1 is a_class_2}
class_1 is a_class_1: {class_1 is a_class_1}
a_class_1 is class_2: {a_class_1 is class_2} 
""")

# retain the old data and does not run __init__
print(class_2.updated)
print(a_class_2.updated)
````
