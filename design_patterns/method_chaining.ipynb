#### Method Chaining

method chaining is useful when u want to allow dev to do multiple different operations together

```python
class MyMethodChain:
    def __init__(self, value):
        self.value = value
    def __str__(self):
        return str(self.value)
    def add(self, value):
        self.value += value
        return self
    def sub(self, value):
        self.value -= value
        return self
    def mul(self, value):
        self.value *= value
        return self
    def div(self, value):
        if value == 0:
            raise ZeroDivisionError
        self.value /= value
        return self

abc = MyMethodChain(10)
abc.mul(2).div(5).add(3).sub(4)
print(abc)
```
