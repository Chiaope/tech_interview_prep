#### Immutable class
Creating an immutable instance is useful so that we prevent the value to be accidentally change

```python
# this method can customise what is being done when item is tried to be set or deleted
class ImmutableRecord:
    def __init__(self, key: str, value: int):
        # Bypass custom __setattr__ to set initial values
        object.__setattr__(self, "key", key)
        object.__setattr__(self, "value", value)

    def __setattr__(self, name, value):
        print("Cannot set")
        raise AttributeError(f"'{type(self).__name__}' object is immutable")

    def __delattr__(self, name):
        print(f"Cannot delete")
        raise AttributeError(f"'{type(self).__name__}' object is immutable")

rec = ImmutableRecord("alpha", 42)
try:
    rec.key = "beta"
except Exception as e:
    print(e)
try:
    del rec.key
except Exception as e:
    print(e)
```

```python
from dataclasses import dataclass

@dataclass(frozen=True)
class MyImmutable:
    x: int
    y: int

mi = MyImmutable(1,2)
print(mi.x,mi.y)
try:
    mi.x = 3
except Exception as e:
    print(e)
try:
    del mi.y
except Exception as e:
    print(e)
```

```python
from typing import NamedTuple

class MyNamedImmutable(NamedTuple):
    x: int
    y: int

mni = MyNamedImmutable(1,2)
print(mni.x,mni.y)
try:
    mni.x = 3
except Exception as e:
    print(e)
try:
    del mni.y
except Exception as e:
    print(e)
```
